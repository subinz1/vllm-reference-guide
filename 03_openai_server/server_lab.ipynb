# Local OpenAI-compatible server lab

Start `serve_local.sh` in a separate terminal. Keep the service on loopback and export `VLLM_API_KEY` only in the active shell.

```python
import os
from openai import OpenAI

client = OpenAI(
    api_key=os.environ['VLLM_API_KEY'],
    base_url=os.environ.get('VLLM_BASE_URL', 'http://127.0.0.1:8000/v1'),
)
models = client.models.list()
[model.id for model in models.data]
```

```python
model = os.environ.get('MODEL_ID', 'Qwen/Qwen2.5-1.5B-Instruct')
response = client.chat.completions.create(
    model=model,
    messages=[{'role': 'user', 'content': 'Explain request queueing in one sentence.'}],
    temperature=0,
)
print(response.choices[0].message.content)
```
