# Offline inference lab

Run this only in a supported vLLM environment with a model that fits the selected device. This is a smoke test, not a throughput benchmark.

```python
from vllm import LLM, SamplingParams

MODEL = 'facebook/opt-125m'  # Replace after checking capacity and licence.
prompts = ['A reproducible serving test records', 'KV cache is important because']
params = SamplingParams(temperature=0.0, max_tokens=32)
llm = LLM(model=MODEL)
outputs = llm.generate(prompts, params)
```

```python
for output in outputs:
    candidate = output.outputs[0]
    print('prompt:', output.prompt)
    print('text:', candidate.text)
    print('finish_reason:', candidate.finish_reason)
    print()
```

## Experiment

Change only `temperature`, rerun, and record the model revision and all request parameters. Do not compare performance with a run that changed model, hardware, or prompt shape.
